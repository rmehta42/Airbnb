```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
file_path = "data/listings.csv"
df = pd.read_csv(file_path)
```

```python
# histogram for 'price'
df = df[df['price'] > 0]  # Remove zero prices to avoid log(0)
df['log_price'] = np.log(df['price'])

plt.figure(figsize=(8, 5))
df['log_price'].hist(bins=50, edgecolor='black')
plt.title("Log Price Distribution")
plt.xlabel("Log(Price)")
plt.ylabel("Frequency")
plt.show()


#2 - 7
#3 - 20
#4 - 54
#5 - 148
#6 - 403
#7 - 1096
#8 - 2980
```

```python
# boxplot for price by neighbourhood
df.boxplot(column='price', by='neighbourhood_group', vert=False, figsize=(14, 6))
plt.title("Price by Neighbourhood Group")
plt.xlabel("Price")
plt.ylabel("Neighbourhood Groups")
plt.xlim(0,800)
plt.show()
```
